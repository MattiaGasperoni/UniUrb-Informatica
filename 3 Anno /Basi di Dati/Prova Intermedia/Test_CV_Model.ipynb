```python
# Installazione libreria e importazione delle dipendenze
!pip install ultralytics
import torch
from ultralytics import YOLO
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload del modello
print("\nCarica il file del modello addestrato (.pt):")
uploaded_model = files.upload()

# Ottenimento del path del modello
model_path = list(uploaded_model.keys())[0]
print(f"Modello caricato: {model_path}")

# Caricamento modello
model = YOLO(model_path)
print(f"Modello pronto per l'inferenza")
print(f"Classi rilevabili: {model.names}")
```

```python
# Upload immagini da testare
print("\nCarica le immagini su cui testare il modello:")
test_images = files.upload()

print(f"\n{len(test_images)} immagini caricate:")
for img_name in test_images.keys():
    print(f"   - {img_name}")
```

```python
print("\n" + "=" * 50)
print("INFERENZA SU IMMAGINI")
print("=" * 50)

# Parametri di inferenza
CONFIDENCE_THRESHOLD = 0.5  # Soglia minima di confidenza
IOU_THRESHOLD        = 0.7  # IoU per Non-Maximum Suppression

# Esegui predizione su ogni immagine
for img_name in test_images.keys():
    print(f"\n{'─'*50}")
    print(f"Elaborazione: {img_name}")
    print(f"{'─'*50}")

    # Predizione
    results = model.predict(
        source=img_name,
        conf=CONFIDENCE_THRESHOLD,
        iou=IOU_THRESHOLD,
        save=False,
        verbose=False
    )

    # Analisi risultati
    for r in results:
        boxes = r.boxes

        # Conta il numero di oggetti rilevati
        num_detections = len(boxes)
        print(f"\nOggetti rilevati: {num_detections}\n")

        if num_detections == 0:
            print(f"   Nessun oggetto rilevato con confidence > {CONFIDENCE_THRESHOLD}")
        else:
            # Stampa dettagli per ogni detection
            for i, box in enumerate(boxes, 1):
                cls_id = int(box.cls[0])
                conf   = float(box.conf[0])

                # Estrazione info del Bounding Box
                xywh     = box.xywh[0].tolist()  # xywh = [x_center, y_center, width, height]
                x_center = xywh[0]
                y_center = xywh[1]
                width    = xywh[2]
                height   = xywh[3]

                print(f"   Oggetto {i}:")
                print(f"      Classe:       {model.names[cls_id]}")
                print(f"      Confidence:   {conf:.2%}")
                print(f"      BBox:         x={x_center:.0f}, y={y_center:.0f}, "
                      f"w={width:.0f}, h={height:.0f}")
                print(f"      Area:         {width * height:.0f} px²\n")

        # Visualizza immagine annotata
        annotated_img = r.plot(
            conf=True,           # Mostra confidence
            labels=True,         # Mostra etichette
            line_width=3,        # Spessore bordi BBox
            font_size=12,        # Dimensione testo
            boxes=True,          # Mostra i box
        )


        print(f"Visualizzazione immagine annotata:\n")
        cv2_imshow(annotated_img)

print(f"\n{'='*50}")
print("INFERENZA COMPLETATA")
print(f"{'='*50}")
```
